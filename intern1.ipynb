```python
import scipy.stats as stats
important_cat_cols = []
for col in dfcat.columns:
    groups = [df['SalePrice'][dfcat[col] == category] for category in dfcat[col].unique()]
    if len(groups) > 1:
        f_stat, p_val = stats.f_oneway(*groups)        
        if p_val < 0.05:
            important_cat_cols.append(col)
dfcat = dfcat[important_cat_cols]
```
